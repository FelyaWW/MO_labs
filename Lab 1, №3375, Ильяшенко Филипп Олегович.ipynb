```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

df = pd.read_csv('FIFA-21 Complete.csv', sep=';')
df.columns = df.columns.str.strip()
print(df.columns.tolist())

num_cols = ['overall', 'age', 'hits', 'potential']

stats_df = df[num_cols].agg(['mean', 'std'])
print('Среднее и стандартное отклонение:\n', stats_df)

# гистограммы
for col in num_cols:
    plt.figure(figsize=(8,4))
    plt.hist(df[col].dropna(), bins=10, alpha=0.7, color='blue', edgecolor='black')
    plt.title(f'Гистограмма распределения {col}')
    plt.xlabel(col)
    plt.ylabel('Частота')
    plt.show()

    z_scores = np.abs(stats.zscore(df[col].dropna()))
    outliers = df[col][z_scores > 3]
    if not outliers.empty:
        print(f'Выбросы в {col}:', outliers.values)
    else:
        print(f'Выбросов в {col} не обнаружено.')

# Проверяем пропуски
missing = df[num_cols].isnull().sum().to_string()
print('Пропущенные значения:\n')
print(missing)
```

```python
# Корреляционная матрица
corr_matrix = df[num_cols].corr()
print("Корреляционная матрица:\n", corr_matrix)

# Визуализация матрицы корреляций через Matplotlib
plt.figure(figsize=(6,5))
plt.imshow(corr_matrix, cmap='coolwarm', interpolation='none')
plt.colorbar(label='Коэффициент корреляции')
plt.xticks(np.arange(len(num_cols)), num_cols, rotation=45)
plt.yticks(np.arange(len(num_cols)), num_cols)
for (i, j), val in np.ndenumerate(corr_matrix):
    plt.text(j, i, f"{val:.2f}", ha='center', va='center', color='black')
plt.title('Матрица корреляций')
plt.tight_layout()
plt.show()

# Графики рассеивания (scatter plot) для каждой пары числовых параметров
for i in range(len(num_cols)):
    for j in range(i+1, len(num_cols)):
        plt.figure(figsize=(5,4))
        plt.scatter(df[num_cols[i]], df[num_cols[j]], alpha=0.7)
        plt.xlabel(num_cols[i])
        plt.ylabel(num_cols[j])
        plt.title(f"{num_cols[i]} vs {num_cols[j]}")
        plt.grid(True)
        plt.show()
```
